```python
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (13, 5)
plt.rcParams['figure.dpi']     = 100

print('Setup complete ')
```

```python
BASE_DIR = Path().resolve().parent
PROC_DIR = BASE_DIR / 'data' / 'processed'

nav   = pd.read_csv(PROC_DIR / 'clean_nav.csv',              parse_dates=['date'])
aum   = pd.read_csv(PROC_DIR / 'clean_aum.csv',              parse_dates=['date'])
sip   = pd.read_csv(PROC_DIR / 'clean_sip.csv',              parse_dates=['month'])
cat   = pd.read_csv(PROC_DIR / 'clean_category_inflows.csv', parse_dates=['month'])
folio = pd.read_csv(PROC_DIR / 'clean_folio_count.csv',      parse_dates=['month'])
txn   = pd.read_csv(PROC_DIR / 'clean_transactions.csv',     parse_dates=['transaction_date'])
port  = pd.read_csv(PROC_DIR / 'clean_portfolio_holdings.csv')
fund  = pd.read_csv(PROC_DIR / 'clean_fund_master.csv')
perf  = pd.read_csv(PROC_DIR / 'clean_performance.csv')

nav_named = nav.merge(fund[['amfi_code','scheme_name','sub_category']], on='amfi_code', how='left')
sip_txn   = txn[txn['transaction_type'] == 'SIP'].copy()

print(f'NAV rows     : {len(nav):,}')
print(f'Transactions : {len(txn):,}')
print(f'Portfolio    : {len(port):,}')
print('All datasets loaded! ')
```

```python
# Chart 1 — NAV Trend Analysis
top10_codes = nav_named['amfi_code'].unique()[:10]
nav10       = nav_named[nav_named['amfi_code'].isin(top10_codes)]

fig = px.line(
    nav10, x='date', y='nav', color='scheme_name',
    title='📈 NAV Trend 2022–2026 — 10 Selected Funds',
    labels={'nav':'NAV (₹)', 'date':'Date', 'scheme_name':'Fund'}
)
fig.add_vrect(x0='2023-01-01', x1='2023-12-31',
              fillcolor='green', opacity=0.08,
              annotation_text='2023 Bull Run',
              annotation_position='top left')
fig.add_vrect(x0='2024-06-01', x1='2024-10-31',
              fillcolor='red', opacity=0.08,
              annotation_text='2024 Correction',
              annotation_position='top left')
fig.update_layout(height=480, legend=dict(font=dict(size=8)))
fig.show()
```

```python
# Chart 2 — AUM Growth by Fund House
aum['year'] = aum['date'].dt.year
aum_yr = aum.groupby(['year','fund_house'])['aum_lakh_crore'].max().reset_index()

fig, ax = plt.subplots(figsize=(14, 6))
sns.barplot(data=aum_yr, x='fund_house', y='aum_lakh_crore', hue='year', ax=ax, palette='Blues')
ax.set_title('AUM Growth by Fund House 2022–2025 (₹ Lakh Crore)', fontsize=14, fontweight='bold')
ax.set_xlabel('Fund House')
ax.set_ylabel('AUM (₹ Lakh Crore)')
ax.tick_params(axis='x', rotation=30)

sbi_max = aum_yr[aum_yr['fund_house']=='SBI Mutual Fund']['aum_lakh_crore'].max()
ax.annotate(f'SBI ₹{sbi_max}L Cr\n(Largest AMC)',
            xy=(0, sbi_max), xytext=(2, sbi_max + 0.5),
            arrowprops=dict(arrowstyle='->', color='red'),
            fontsize=9, color='red', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Chart 3 — SIP Inflow Time-Series
fig = px.line(
    sip, x='month', y='sip_inflow_crore',
    title='📈 Monthly SIP Inflow Jan 2022 – Dec 2025',
    labels={'sip_inflow_crore':'SIP Inflow (₹ Crore)', 'month':'Month'},
    markers=True
)
ath_row = sip.loc[sip['sip_inflow_crore'].idxmax()]
fig.add_annotation(
    x=ath_row['month'], y=ath_row['sip_inflow_crore'],
    text=f"All-Time High<br>₹{int(ath_row['sip_inflow_crore']):,} Cr",
    showarrow=True, arrowhead=2, arrowcolor='red',
    font=dict(color='red', size=11), bgcolor='lightyellow'
)
fig.update_traces(line_color='royalblue', line_width=2)
fig.update_layout(height=430)
fig.show()
```

```python
# Chart 4 — Category Inflow Heatmap
cat['month_str'] = cat['month'].dt.strftime('%Y-%m')
pivot = cat.pivot_table(index='category', columns='month_str',
                        values='net_inflow_crore', aggfunc='sum')

fig, ax = plt.subplots(figsize=(18, 6))
sns.heatmap(pivot, cmap='RdYlGn', linewidths=0.3, ax=ax,
            cbar_kws={'label':'Net Inflow (₹ Crore)'})
ax.set_title('Category-wise Net Inflow Heatmap', fontsize=14, fontweight='bold')
ax.set_xlabel('Month')
ax.set_ylabel('Category')
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Chart 5 — Investor Demographics (FIXED)
age_order  = ['18-25','26-35','36-45','46-55','56+']
age_counts = sip_txn['age_group'].value_counts().reindex(age_order).fillna(0)

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# 5a — Age group pie
axes[0].pie(age_counts, labels=age_counts.index, autopct='%1.1f%%',
            startangle=140, colors=sns.color_palette('pastel', 5))
axes[0].set_title('Age Group Distribution', fontweight='bold')

# 5b — SIP amount box plot by age group
sip_txn['age_group'] = pd.Categorical(sip_txn['age_group'],
                        categories=age_order, ordered=True)
sns.boxplot(data=sip_txn.sort_values('age_group'),
            x='age_group', y='amount_inr',
            palette='muted', ax=axes[1])
axes[1].set_title('SIP Amount by Age Group', fontweight='bold')
axes[1].set_xlabel('Age Group')
axes[1].set_ylabel('SIP Amount (₹)')
axes[1].yaxis.set_major_formatter(
    mticker.FuncFormatter(lambda x,_: f'₹{x:,.0f}'))

# 5c — Gender split pie
gender_counts = sip_txn['gender'].value_counts()
axes[2].pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%',
            startangle=90, colors=['#AEC6CF','#FFB347'])
axes[2].set_title('Gender Split', fontweight='bold')

plt.suptitle('Investor Demographics Analysis', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Chart 6 — Geographic Distribution
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 6a — SIP amount by state horizontal bar
state_sip = (sip_txn.groupby('state')['amount_inr']
             .sum().sort_values(ascending=True) / 1e7)
colors = sns.color_palette('Blues_d', len(state_sip))
axes[0].barh(state_sip.index, state_sip.values, color=colors)
axes[0].set_title('Total SIP Investment by State (₹ Crore)', fontweight='bold')
axes[0].set_xlabel('Total SIP Amount (₹ Crore)')
for i, val in enumerate(state_sip.values):
    axes[0].text(val + 0.5, i, f'₹{val:.0f}Cr', va='center', fontsize=8)

# 6b — T30 vs B30 pie
tier_counts = sip_txn['city_tier'].value_counts()
axes[1].pie(tier_counts, labels=tier_counts.index, autopct='%1.1f%%',
            startangle=90, colors=['#4C72B0','#DD8452'])
axes[1].set_title('T30 vs B30 City Tier — SIP Transactions', fontweight='bold')

plt.suptitle('Geographic Distribution of SIP Investments', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Chart 7 — Folio Count Growth
fig = px.line(
    folio, x='month', y='total_folios_crore',
    title='📂 Industry Folio Count Growth (Jan 2022 – Dec 2025)',
    labels={'total_folios_crore':'Total Folios (Crore)', 'month':'Month'},
    markers=True, line_shape='spline'
)
fig.add_annotation(x=folio['month'].iloc[0], y=folio['total_folios_crore'].iloc[0],
                   text='13.26 Cr<br>Jan 2022', showarrow=True, arrowhead=2,
                   font=dict(color='blue', size=10))
fig.add_annotation(x=folio['month'].iloc[-1], y=folio['total_folios_crore'].iloc[-1],
                   text='26.12 Cr<br>Dec 2025', showarrow=True, arrowhead=2,
                   font=dict(color='green', size=10))
milestone = folio[folio['total_folios_crore'] >= 20].iloc[0]
fig.add_annotation(x=milestone['month'], y=milestone['total_folios_crore'],
                   text='20 Cr Milestone!', showarrow=True, arrowhead=2,
                   font=dict(color='orange', size=10))
fig.update_traces(line_color='teal', line_width=2)
fig.update_layout(height=430)
fig.show()
```

```python
# Chart 8 — NAV Return Correlation Matrix
ten_codes = nav_named['amfi_code'].unique()[:10]
names_map = (nav_named[nav_named['amfi_code'].isin(ten_codes)]
             [['amfi_code','scheme_name']].drop_duplicates()
             .set_index('amfi_code')['scheme_name'].str[:18].to_dict())

nav_pivot = (nav[nav['amfi_code'].isin(ten_codes)]
             .pivot_table(index='date', columns='amfi_code', values='nav')
             .pct_change())
nav_pivot.columns = [names_map.get(c, str(c)) for c in nav_pivot.columns]
corr = nav_pivot.corr()

fig, ax = plt.subplots(figsize=(12, 9))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
            vmin=-1, vmax=1, linewidths=0.5, ax=ax,
            cbar_kws={'shrink':0.8})
ax.set_title('NAV Daily Return Correlation Matrix (10 Funds)',
             fontsize=13, fontweight='bold')
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Chart 9 — Sector Allocation Donut
sector_wt = (port.groupby('sector')['weight_pct']
             .sum().sort_values(ascending=False).reset_index())
sector_wt['pct'] = sector_wt['weight_pct'] / sector_wt['weight_pct'].sum() * 100

fig = px.pie(
    sector_wt, names='sector', values='pct', hole=0.45,
    title='🍩 Sector Allocation Across All Equity Funds (%)',
    color_discrete_sequence=px.colors.qualitative.Set3
)
fig.update_traces(textposition='outside', textinfo='percent+label')
fig.update_layout(height=550)
fig.show()
```

```python
# Chart 10 — Top 10 Funds by 3-Year Return
top10_ret = perf.nlargest(10, 'return_3yr_pct')

fig, ax = plt.subplots(figsize=(11, 6))
bars = ax.barh(top10_ret['scheme_name'].str[:30], top10_ret['return_3yr_pct'],
               color=sns.color_palette('viridis', 10))
for bar, val in zip(bars, top10_ret['return_3yr_pct']):
    ax.text(bar.get_width()+0.2, bar.get_y()+bar.get_height()/2,
            f'{val:.1f}%', va='center', fontsize=9)
ax.set_title('Top 10 Funds by 3-Year CAGR (%)', fontsize=13, fontweight='bold')
ax.set_xlabel('3-Year CAGR (%)')
plt.tight_layout()
plt.show()
```

```python
# Chart 11 — Risk vs Return Scatter
fig = px.scatter(
    perf, x='std_dev_ann_pct', y='return_3yr_pct',
    size='aum_crore', color='risk_grade', hover_name='scheme_name',
    title=' Risk vs Return — All 40 Funds (Bubble Size = AUM)',
    labels={'std_dev_ann_pct':'Risk / Std Dev (%)','return_3yr_pct':'3-Year Return (%)'}
)
fig.update_layout(height=500)
fig.show()
```

```python
# Chart 12 — Monthly Transaction Volume by Type
txn['month'] = txn['transaction_date'].dt.to_period('M').dt.to_timestamp()
txn_vol = txn.groupby(['month','transaction_type'])['amount_inr'].sum().reset_index()

fig = px.line(
    txn_vol, x='month', y='amount_inr', color='transaction_type',
    title='📅 Monthly Transaction Volume by Type (SIP / Lumpsum / Redemption)',
    labels={'amount_inr':'Total Amount (₹)','month':'Month'}
)
fig.update_layout(height=430)
fig.show()
```

```python
# Chart 13 — AUM Market Share by Fund House
latest_aum = aum[aum['date'] == aum['date'].max()]

fig = px.pie(
    latest_aum, names='fund_house', values='aum_lakh_crore',
    title=f'AUM Market Share — {aum["date"].max().strftime("%b %Y")}',
    hole=0.3, color_discrete_sequence=px.colors.qualitative.Plotly
)
fig.update_layout(height=480)
fig.show()
```

```python
# Chart 14 — Sharpe Ratio Ranking
sharpe_sorted = perf.sort_values('sharpe_ratio', ascending=True)
colors = ['#2ecc71' if x > 1 else '#e74c3c' if x < 0 else '#f39c12'
          for x in sharpe_sorted['sharpe_ratio']]

fig, ax = plt.subplots(figsize=(10, 13))
ax.barh(sharpe_sorted['scheme_name'].str[:30], sharpe_sorted['sharpe_ratio'],
        color=colors)
ax.axvline(x=1, color='green', linestyle='--', linewidth=1.2, label='Sharpe=1 (Good)')
ax.axvline(x=0, color='red',   linestyle='--', linewidth=1.2, label='Sharpe=0')
ax.set_title('Sharpe Ratio Ranking — All 40 Funds', fontsize=13, fontweight='bold')
ax.set_xlabel('Sharpe Ratio')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Chart 15 — SIP Inflow vs Active Accounts (Dual Axis)
fig, ax1 = plt.subplots(figsize=(13, 5))
ax2 = ax1.twinx()

ax1.bar(sip['month'], sip['sip_inflow_crore'],
        color='steelblue', alpha=0.6, label='SIP Inflow (₹ Cr)', width=20)
ax2.plot(sip['month'], sip['active_sip_accounts_crore'],
         color='orange', linewidth=2.5, marker='o', label='Active Accounts (Cr)')

ax1.set_title('SIP Inflow vs Active SIP Accounts Growth',
              fontsize=13, fontweight='bold')
ax1.set_ylabel('SIP Inflow (₹ Crore)', color='steelblue')
ax2.set_ylabel('Active SIP Accounts (Crore)', color='orange')
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
## 10 Key EDA Findings

**Finding 1 — 2023 Bull Run:** All 40 equity funds showed strong NAV appreciation during 2023 with average returns exceeding 25%. *(Chart 1)*

**Finding 2 — SBI Dominance:** SBI Mutual Fund leads AUM at ₹12.5 lakh crore, 1.2× larger than second-placed ICICI Prudential. *(Chart 2)*

**Finding 3 — SIP All-Time High:** Monthly SIP inflows hit ₹31,002 crore in Dec 2025 — the all-time high. *(Chart 3)*

**Finding 4 — Mid Cap Leads Inflows:** Mid Cap and Small Cap categories received highest net inflows in FY 2024–25. *(Chart 4)*

**Finding 5 — Young Investors Lead:** The 26–35 age group accounts for highest share of SIP investors. *(Chart 5)*

**Finding 6 — Maharashtra Tops States:** Maharashtra and Karnataka contribute highest SIP investment volumes. *(Chart 6)*

**Finding 7 — Folios Doubled:** Industry folios grew from 13.26 Cr to 26.12 Cr — doubling in 4 years. *(Chart 7)*

**Finding 8 — High Fund Correlation:** Most large-cap equity fund returns show correlation above 0.85. *(Chart 8)*

**Finding 9 — Banking Dominates Portfolios:** Banking and Financial Services has highest portfolio allocation 25%+. *(Chart 9)*

**Finding 10 — Risk-Return Tradeoff Confirmed:** Small Cap funds show higher risk but also deliver higher 3-year CAGR vs Large Cap. *(Chart 11)*
```
